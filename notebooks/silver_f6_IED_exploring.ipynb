```python
query_f6_be_functionnal="""SELECT * FROM bronze.f6_1_ied_installations
WHERE "CountryName"='Belgium' AND "installationStatus"='functional'

"""
```

```python
f6_be_functional=pd.read_sql(query_f6_be_functionnal,con=engine)
```

```python
f6_1_col_to_keep=['reportingYear',
       'parent_facilityInspireId',
       'InstallationInspireId',
       'City_of_Facility',
       'Longitude', 'Latitude',
       ]
```

```python
f6_be_functional=f6_be_functional[f6_1_col_to_keep].copy()
```

```python
f6_be_functional=f6_be_functional.drop_duplicates()
```

```python
f6_be_functional.head()
```

```python
f6_be_functional.isnull().sum()
```

```python
f6_be_functional.info()
```

```python
f6_be_functional=f6_be_functional.rename(str.lower, axis='columns')
```

```python
def normalize_text(x):
    if pd.isna(x):
        return None
    x = str(x).lower().strip()
    for ch in ["-", "/", "(", ")", ",", ".", ";", ":", "'"]:
        x = x.replace(ch, " ")
    x = " ".join(x.split())
    return x
```

```python
f6_be_functional['city_of_facility']=f6_be_functional['city_of_facility'].map(normalize_text)
```

```python
f6_be_functional.head()
```

```python
f6_be_functional.isnull().sum()
```

```python
f6_be_functional=f6_be_functional.fillna(value={'city_of_facility':'unknown'})
```
